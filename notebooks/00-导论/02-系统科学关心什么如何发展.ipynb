## 同一个现象，可以引出一组不同的问题

继续考虑运动后的恢复。可以问身体接下来怎样变化，也可以问怎样从有限读数认识内部情况，还可以问怎样改变环境来影响恢复。目的不同，需要的信息与方法也会不同。

本书借助郭雷《系统学是什么》的“五论”组织这些问题。下面的通俗提问与篇章联系是本书的学习安排；五类问题在一次研究中常常同时出现。[参考原文](https://sss.bnu.edu.cn/xtzc/xtkp/fec3b17379c3432b8d19654c8b4817a5.htm)

| 视角 | 用自己的话提问 | 恢复案例中的线索 |
|---|---|---|
| 系统方法论（systems methodology） | 怎样组织问题、假设与证据？ | 研究什么范围，用什么结果比较解释？ |
| 系统演化论（theory of system evolution） | 状态（state）、结构与行为怎样在时空中改变？ | 恢复会趋于平稳、出现振荡，还是发生更复杂的变化？ |
| 系统认知论（theory of system cognition） | 怎样从观测（observation）、机制与先验（prior）知识认识系统？ | 体表读数能够告诉我们哪些内部情况？ |
| 系统调控论（theory of system regulation） | 怎样通过行动改变系统行为？ | 能改变哪些条件，需要权衡什么？ |
| 系统实践论（systems practice） | 怎样在具体对象上实施并检验方法？ | 新实验是否支持预测，效果是否满足实际目的？ |

## 从问题看发展脉络

### 跨越对象，寻找共同的组织关系

研究者长期分别研究力学、生命和工程中的变化问题。一般系统论（general system theory，GST）进一步提出：生物、机器和社会组织中，能否比较组成之间的关系，并建立可跨领域讨论的模型（model）？Bertalanffy 的 [1950 年论文](https://www.journals.uchicago.edu/doi/10.1093/bjps/I.2.134)是理解这一传统的代表性入口。开放系统与环境持续交换的视角，帮助我们理解生命系统为何需要把外界条件纳入研究。

### 调节离不开信息，信息也需要被度量

如果观察到变化后采取行动，行动又改变之后的观察，就形成了反馈（feedback）。控制论（cybernetics）把动物与机器中的控制、通信和调节放在共同的问题下讨论。Wiener 的 [《控制论》](https://mitpress.mit.edu/9780262537841/cybernetics-or-control-and-communication-in-the-animal-and-the-machine/)初版于 1948 年，为这一研究传统提供了重要表达。

同年，Shannon 的 [《通信的数学理论》](https://www.nokia.com/bell-labs/publications-and-media/publications/a-mathematical-theory-of-communication/)发展了信息论（information theory）的定量框架，研究不确定性和通信传递的限制。这使“获得多少信息”和“传递受什么限制”成为可计算的问题；消息的生理含义仍要结合具体对象解释。第一篇会正式建立反馈概念，第五篇再引入概率（probability）与信息的数学。

### 规则确定，长期行为仍可能难以预测

动力系统（dynamical system）研究状态怎样随时间演化。非线性动力学（nonlinear dynamics）让我们研究多种长期行为及其变化。Lorenz 的 [1963 年工作](https://doi.org/10.1175/1520-0469%281963%29020%3C0130%3ADNF%3E2.0.CO%3B2)展示了简单确定性方程中的非周期（aperiodic）行为与预测困难，成为混沌（chaos）研究的经典案例。规则明确，并不保证任意远的未来都能准确预测；第 4 篇会学习怎样区分这种困难、随机性和计算误差。

### 从相互作用，到集体现象与数据融合

复杂系统（complex system）研究中的重要问题是：许多部分相互作用，怎样形成集体行为？网络科学（network science）帮助描述连接结构，多主体模型（agent-based model，ABM）帮助研究局部规则怎样形成群体结果。第 11—13 篇会把传播、同步、演化与韧性变成可计算的问题。

随着计算与观测手段发展，数据驱动建模（data-driven modeling）为发现动态关系提供了更多方法；已知机制又能约束学习结果。第 14—17 篇将以方程发现、神经微分方程（neural differential equation）等具体工作讨论这种融合，并用传统方法作比较。相关原始文献随[对应篇章](../../docs/教材设计.md#references)给出。

## 中国系统科学的一条相关脉络

钱学森的《工程控制论（engineering cybernetics）》（Engineering Cybernetics，1954）连接了工程中的动态分析与控制问题；之后对系统工程和系统学的推动，又把理论、技术与实践的关系纳入讨论。中国系统科学（systems science）的学科建设，以及郭雷在 2016 年对“五论”的讨论，为本书组织这些问题提供了中文语境中的参照。[郭雷的回顾与讨论](https://sss.bnu.edu.cn/xtzc/xtkp/fec3b17379c3432b8d19654c8b4817a5.htm)

下图把代表性工作放在不同线上，右侧汇入“今天的问题”。年份只帮助定位文献；连线表示本节讨论的问题联系，并不声称某篇论文单独创立了整个领域。图中的 General systems、Control & information、Dynamics、Complexity & data 和 Systems practice 分别对应共同组织、控制与信息、动态行为、复杂性与数据、系统实践。Tsien 指钱学森，Guo 指郭雷。

```python
import matplotlib.pyplot as plt
from systems_science import plot_style

plot_style()
fig, ax = plt.subplots(figsize=(11, 5))
ax.set(xlim=(0, 11), ylim=(-0.4, 5.1))
ax.axis('off')
rows = [
    ('General systems', 'Bertalanffy, 1950', 4.4),
    ('Control & information', 'Wiener / Shannon, 1948', 3.4),
    ('Dynamics', 'Lorenz, 1963', 2.4),
    ('Complexity & data', 'Networks / learning', 1.4),
    ('Systems practice', 'Tsien, 1954 / Guo, 2016', 0.4),
]
destination = ax.text(9.4, 2.4, 'Questions\ntoday', ha='center', va='center', fontsize=13, zorder=3,
        bbox=dict(boxstyle='round,pad=0.8', facecolor='white', edgecolor='#14476f'))
works = []
for label, work, y in rows:
    ax.text(0.1, y, label, va='center', fontsize=12)
    works.append(ax.text(4.5, y, work, ha='center', va='center', fontsize=11, zorder=3,
            bbox=dict(boxstyle='round,pad=0.45', facecolor='white', edgecolor='#6491b5')))
fig.canvas.draw()
for (_, _, y), work in zip(rows, works):
    ax.annotate('', xy=(9.4, 2.4), xytext=(4.5, y),
                arrowprops=dict(arrowstyle='->', color='#4078a0', shrinkA=5, shrinkB=5,
                                patchA=work.get_bbox_patch(), patchB=destination.get_bbox_patch()), zorder=2)
ax.set_title("Intersecting traditions and today's questions", pad=12)
plt.show()
```

## 怎样使用这幅全景

学习控制方法时，仍需要知道对象怎样演化、读数告诉了我们什么、结论如何验证。学习数据方法时，仍需要理解机制和观测条件。一项研究可以同时提出几类问题；选方法时先看目的和证据，再看方法名称。

回到恢复案例：预测需要演化关系，从读数推断内部情况需要认识方法，选择行动需要调控与优化，实际效果需要新实验。下一节用一个更小的传播问题，让我们完整走一遍“提出问题—写清规则—计算—检查结论”的过程。
