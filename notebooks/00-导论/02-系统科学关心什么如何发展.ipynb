## 有了温度曲线，我们还缺少什么？

上一节，我们把运动后的产热、热量运输和散热放在一起，提出了一个恢复问题。假设现在已经记录到一条体表温度曲线，接下来就能解释身体的恢复了吗？

先想想这条曲线告诉了我们什么。它是一组观测（observation），记录了某处在不同时刻测得的温度。要了解体内的情况，还需要知道被测位置与内部热量变化之间的关系；要预测以后几分钟，还需要知道这些过程怎样继续；要改变恢复过程，则需要选择能采取的行动。虽然围绕同一条曲线，这些任务仍有不同的困难。

一个模型（model）可以把所考虑的关系写成明确规则。例如，我们可以表示热量如何从内部到达体表，再表示它如何向环境散失。这样的规则使读数和过程之间有了可检查的联系。模型怎样建立、测量怎样解释、行动怎样选择，又分别需要不同的方法。系统科学（systems science）的发展，就与研究者不断面对这些困难有关。


## 观察怎样变成行动？

先看一个人工温控装置。我们希望温度接近目标，因此让它定期读取温度；如果读数高于目标，就增强冷却，之后再测一次。注意，这里的测量不只是记录：它决定了动作，动作又改变了下一次测量到的温度。沿着“温度变化 → 读取温度 → 调整冷却 → 温度再次变化”走一圈，就能看见反馈（feedback）：作用的结果经过这条联系返回，影响后续作用。

你也许会觉得，只要规定“热了就冷却”，问题便解决了。可如果温度计偏高，装置可能一直冷却得过多；如果读取的是较早的温度，到达目标后也可能继续冷却。要研究调节效果，我们既要了解动作怎样影响对象，也要了解动作依据的信息是否可靠、是否及时。读取与动作之间相隔的时间形成了延迟（delay）；第一篇会用一个能手算的小例子解释它怎样影响调节。

控制论（cybernetics）把动物和机器中的控制、通信与调节放在共同的问题下研究。Wiener 于 1948 年出版的 [《控制论》](https://mitpress.mit.edu/9780262537841/cybernetics-or-control-and-communication-in-the-animal-and-the-machine/)是这一传统的重要著作。它使我们能够比较身体调节与机器控制中的共同关系，同时继续研究各自具体的作用机制。

信息是否及时和可靠，也提出了另一组问题。传输通道能送来多少信息？干扰会使我们丢失多少区分不同情况的能力？Shannon 在 1948 年的 [《通信的数学理论》](https://www.nokia.com/bell-labs/publications-and-media/publications/a-mathematical-theory-of-communication/)中建立了信息论（information theory）的定量框架，研究通信及其限制。这里衡量信息，是在讨论不确定性和传递能力；一条消息对人体有什么生理意义，还需要由研究对象来解释。温度读数即使传输得毫无差错，也仍可能因被测位置不合适而回答不了我们的问题。


## 规则明确，就能一直预测下去吗？

有了当前的情况和作用规则，我们自然会想把未来算出来。在一个模型中，为了描述当前情况并继续计算，我们需要保留一些量。例如，研究两个部位之间的热量交换，可能需要同时保留两处的温度，而不能只留下它们的平均值。这些用来描述模型当前情况的量称为状态（state）。动力系统（dynamical system）研究状态怎样随时间变化。

状态的变化可以趋于平稳，也可以反复振荡。有时，一个小变化的作用还取决于它发生在什么范围：同样改变一点温度，所触发的调节作用未必总按同一比例改变。非线性动力学（nonlinear dynamics）研究这类关系及它们造成的动态行为，帮助我们解释为什么一种规则会出现多种长期结果。

这里还有一个反直觉的问题：规则完全确定，为什么预测仍可能失败？Lorenz 在 [1963 年的论文](https://doi.org/10.1175/1520-0469%281963%29020%3C0130%3ADNF%3E2.0.CO%3B2)中研究了一个简化的对流模型。起点略有差别的计算过程，后来可以出现很大差别。这成为混沌（chaos）研究的经典案例：起点测得不够精确，就可能限制我们能预测多远。

因此，知道规则与知道未来之间，还有一段要研究的距离。我们需要问起点知道得有多准、计算误差会怎样积累，以及预测要持续多长时间。第四篇会把这些问题分别展开，并区分由确定规则产生的预测困难与随机变化。它们不因曲线看起来都不规则，就成为同一种现象。


## 为什么不同对象能共享一些方法？

现在把人体、人工温控装置和消息传递放在一起。它们传递的东西不同：人体交换热量与物质，温控装置利用测量来决定动作，人群沿联系传递消息。可是，我们都要研究哪些部分联系起来、联系怎样传递影响、环境怎样改变结果。相同的研究问题，使不同领域的方法有机会互相启发。

一般系统论（general system theory，GST）关注跨越具体对象的组织与相互作用关系。Bertalanffy 的 [1950 年论文《一般系统论纲要》](https://www.journals.uchicago.edu/doi/10.1093/bjps/I.2.134)是这一传统的代表性文献。研究生命系统时，把与环境的持续交换放进来，也让我们看到：对象能够维持某种整体行为，往往与内部过程和外界条件共同作用有关。

如果组成很多，又会出现新的困难。我们知道每个人怎么传消息，却还未必知道全体多久收到；我们知道个体怎样调整行为，却还需要研究这种调整会不会形成群体的同步、竞争或合作。复杂系统（complex system）研究关注许多部分相互作用形成的集体现象。网络科学（network science）帮助表示谁与谁连接；多主体模型（agent-based model，ABM）则通过个体规则计算群体结果。下一节只用六个人，便能看见连接安排怎样改变传播过程。

观测与计算手段的发展又提供了另一条道路：当某些规则难以直接写出时，可以尝试从数据中建立它们。数据驱动建模（data-driven modeling）由此进入动态关系的研究。但如果预测到的物质变化违背了已知的流入、流出关系，就需要检查学到的规则。因此，数据与机制经常要一起使用：已有机制约束可接受的解释，新数据检查这些解释，并帮助发现缺少的关系。第十四至第十七篇会以方程发现、神经微分方程（neural differential equation）等具体方法展开这项研究，所用原始文献见[相应参考表](../../docs/教材设计.md#references)。


## 把方法带回实际问题

一套方法从理论走到实际对象，还要面对测量条件、可采取的动作和真实约束。在中国系统科学的发展中，钱学森把动态分析、工程控制与系统工程联系起来，推动了理论、技术和实践之间的讨论。他的《工程控制论（engineering cybernetics）》出版于 1954 年，研究工程控制系统的分析与设计；出版年份与内容可对照[加州理工学院的介绍](https://aerospace.caltech.edu/about/legends-of-galcit/qian-xuesen-tsien-hsue-shen)。

郭雷在 2016 年的[《系统学是什么》](https://tcct.amss.ac.cn/sc-journal/2016/201601/20160104.pdf)中，用“五论”讨论系统学的主要问题，并强调它们彼此联系。经过前面的例子，我们可以把这些名称整理成下面这张地图。表中的提问是本书对学习任务的说明。

| 研究视角 | 在恢复问题中可以怎样提问？ |
|---|---|
| 系统方法论（systems methodology） | 先研究哪些部分？怎样用假设和证据组织研究？ |
| 系统演化论（theory of system evolution） | 温度和调节过程怎样随时间变化？ |
| 系统认知论（theory of system cognition） | 体表读数能告诉我们哪些内部情况？ |
| 系统调控论（theory of system regulation） | 改变哪些条件，才能接近给定目标？ |
| 系统实践论（systems practice） | 怎样在具体对象上实施方法，检查实际效果？ |

例如，研究者先根据读数推断内部情况，再选择风扇风速，最后在一次新实验里检查预测。这三个步骤分别突出认识、调控和实践核验，但它们始终相连：读数的解释影响动作，动作依赖变化规律，实际结果又会反过来检验原来的解释。

下面的图把本节提到的几条学术脉络连接到共同的研究问题。年份标明代表性文献的位置；这些传统经过长期发展，也不断互相影响。


```python
import matplotlib.pyplot as plt
from systems_science import plot_style

plot_style()
plt.rcParams['font.sans-serif'] = ['Microsoft YaHei', 'SimHei', 'Noto Sans CJK SC', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False
fig, ax = plt.subplots(figsize=(11, 5))
ax.set(xlim=(0, 11), ylim=(-0.4, 5.1))
ax.axis('off')
rows = [
    ('组织与相互作用', 'Bertalanffy，1950', 4.4),
    ('控制与信息', 'Wiener / Shannon，1948', 3.4),
    ('动态行为与预测', 'Lorenz，1963', 2.4),
    ('集体现象与数据', '网络 / 个体规则 / 学习', 1.4),
    ('理论与实践', '钱学森，1954 / 郭雷，2016', 0.4),
]
destination = ax.text(9.4, 2.4, '理解、预测\n与调节系统', ha='center', va='center', fontsize=13, zorder=3,
        bbox=dict(boxstyle='round,pad=0.8', facecolor='white', edgecolor='#14476f'))
works = []
for label, work, y in rows:
    ax.text(0.1, y, label, va='center', fontsize=12)
    works.append(ax.text(4.5, y, work, ha='center', va='center', fontsize=11, zorder=3,
            bbox=dict(boxstyle='round,pad=0.45', facecolor='white', edgecolor='#6491b5')))
fig.canvas.draw()
for (_, _, y), work in zip(rows, works):
    ax.annotate('', xy=(9.4, 2.4), xytext=(4.5, y),
                arrowprops=dict(arrowstyle='->', color='#4078a0', shrinkA=5, shrinkB=5,
                                patchA=work.get_bbox_patch(), patchB=destination.get_bbox_patch()), zorder=2)
ax.set_title('多条学术脉络，共同回应系统问题', pad=12)
plt.show()
```

研究方法的分工，最终要由问题来决定。只有一条温度曲线时，我们要补足读数与对象之间的关系；想预测下一段时，要建立演化规则；想改变过程时，要考虑动作与结果怎样联系；得到新结果以后，再检查原来的判断。学科名称为这些方法提供了来历，实际研究则要把它们接起来。
