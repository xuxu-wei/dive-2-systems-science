## 停下脚步，身体为什么还在恢复？

跑步结束后，我们已经站着不动，心跳和呼吸却还要过一段时间才能平缓下来。身体的恢复并没有随着一个动作的结束而立刻完成。要解释这段过程，可以从肌肉、心脏和皮肤各自的作用入手。可是一份器官名单还没有告诉我们：这些作用怎样共同形成一次恢复过程？

先把问题缩小到热量。运动时，肌肉活动产生热量；循环把热量带到身体的不同部位；热量又通过皮肤等途径向环境散失。停止运动改变了产热的情况，身体已经积累的热量却仍需要通过交换才能散去。我们想解释的，正是这些过程共同作用后的变化。

这里有一个容易忽略的细节：汗液分泌出来以后，还要蒸发才能通过这种途径带走热量。空气的湿度和流动会影响蒸发，所以出汗量相近的两次活动，散热过程也可能不同。身体内部还会调节皮肤血流和出汗。解释温度怎样变化，需要把内部调节和环境条件一起考虑。有关体温调节的生理依据，可阅读 [OpenStax 的稳态一节](https://openstax.org/books/anatomy-and-physiology-2e/pages/1-5-homeostasis)。

不妨先停在这里想一想：如果两次活动后的温度曲线不同，我们只记录了活动和温度，却没有记录环境条件，还能判断差异来自哪里吗？我们甚至还不知道，测到的是体表哪一处的温度。这些问题会决定下一步该记录什么。


## 把过程连接起来

下面的图把刚才提到的过程放在一起。上方从活动和产热出发，经过体内热量与循环运输，连接到皮肤和蒸发散热；下方补上身体调节与环境条件。沿箭头走，可以看见为什么一个局部过程的变化会影响另一个过程。例如，皮肤血流改变了热量到达体表的过程，环境条件又影响热量如何离开体表。

这张图先帮助我们提出问题。箭头说明考虑了哪类影响，其强弱和具体发生过程还需要测量与解释。


```python
import matplotlib.pyplot as plt
from systems_science import plot_style

plot_style()
plt.rcParams['font.sans-serif'] = ['Microsoft YaHei', 'SimHei', 'Noto Sans CJK SC', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False
fig, ax = plt.subplots(figsize=(10, 4.5))
ax.set(xlim=(0, 10), ylim=(0, 5))
ax.axis('off')
positions = [(1.5, 3.5), (5, 3.5), (8.5, 3.5), (3.5, 1.1), (8.5, 1.1)]
labels = ['活动与产热', '体内热量\n与循环运输', '皮肤与\n蒸发散热', '身体调节', '环境条件']
boxes = []
for (x, y), label in zip(positions, labels):
    boxes.append(ax.text(x, y, label, ha='center', va='center', fontsize=12, zorder=3,
            bbox=dict(boxstyle='round,pad=0.6', facecolor='#ffffff', edgecolor='#6491b5')))
fig.canvas.draw()
# 箭头表示本图考虑的影响关系，没有设定影响强度。
for source, target in [(0, 1), (1, 2), (1, 3), (3, 1), (3, 2), (4, 2)]:
    ax.annotate('', xy=positions[target], xytext=positions[source],
                arrowprops=dict(arrowstyle='->', color='#4078a0',
                                patchA=boxes[source].get_bbox_patch(),
                                patchB=boxes[target].get_bbox_patch(),
                                shrinkA=5, shrinkB=5, connectionstyle='arc3,rad=0.08'), zorder=2)
ax.set_title('恢复问题：过程之间怎样联系？', pad=12)
plt.show()
```

我们把相互关联、相互作用的组成部分所构成的整体称为**系统（system）**。在恢复问题里，肌肉、循环和皮肤等部分通过产热、运输和散热联系起来。我们关心的是这些联系共同造成的温度变化；周围空气则是与身体交换热量的环境。

选择研究什么整体，取决于要回答的问题。如果想解释体内不同位置的温度，可能需要区分身体内部与体表；如果想解释室内温度为什么上升，还要把人体向房间散失的热量接入房间的研究。对象的名称没有替我们完成这项选择。提出问题以后，我们才知道哪些部分、哪些联系和哪些环境条件需要放进来。

**系统科学（systems science）**研究的正是组成、相互作用、环境与整体行为之间的关系，并发展理解、预测和调节这些行为的方法。刚才，我们从“身体怎样恢复”走到“哪些过程通过什么联系共同影响恢复”，就已经开始使用这种研究视角。

这也解释了为什么系统科学经常跨越学科。身体与恒温装置都可能根据温度变化调整散热，我们可以比较它们如何利用当前情况采取行动。不过，比较之前仍要了解各自的组成和作用方式。恒温装置的风扇与人的汗腺怎样工作，需要不同的领域知识；共同的关系则帮助我们提出相似的研究问题。


## 一个读数下降，可以有几种解释？

把目光转到体内物质。假设我们追踪一种物质，发现它在血液中的读数下降。一个很自然的解释是：“它已经被身体清除了。”但还有一种可能：物质从血液转移到了组织，仍然留在身体里。只看血液中的读数，这两个过程可能产生相似的变化。

如果研究血液这一部分，进入组织的物质确实离开了我们的观察范围；如果研究整个身体，它只是改变了位置。于是，同一个局部读数，对不同研究问题有不同含义。我们需要知道物质在哪里测得、哪些位置之间可以交换，以及什么过程真正使它离开整体。

为了计算这些变化，后面会建立模型（model）：用明确的变量和规则表示当前要研究的关系。模型可以告诉我们，假如某些交换和清除规则成立，读数应该怎样变化；新的观测（observation），也就是实际记录到的量和情况，再帮助我们检查这些预测。局部机制为模型提供依据，预测与观测的差异又会提示还缺少哪一项机制或测量。

组成之间的联系还可能直接改变整体行为。六个人排成一条消息传递链，与一个人连接其余五个人，虽然人数和联系数相同，消息到达各人的过程却可能不同。你可以先试着从第一个人沿联系走一步。第三节会把这个问题算清楚。


## 我们究竟想知道什么？

现在回到运动后的恢复。问“温度怎样变化”，首先需要记录温度所在的位置、时刻和环境，这些记录帮助我们描述过程。问“为什么这样变化”，还需要比较产热、运输和散热等解释，寻找能够区分它们的证据。仅有一条下降的曲线，通常还不足以完成这一步。

如果问“接下来几分钟会怎样”，就需要一个能够产生预测的模型。我们可以用前一段记录建立它，再与后一段新记录比较。这里，新记录有特殊作用：模型已经见过的数据能帮助建立规则，模型尚未见过的数据则能检查它是否解释了新的情况。

如果问“增大风速会带来什么变化”，我们又增加了行动。除了温度怎样变化，还要考虑可改变的条件、希望达到的目标和动作的代价。例如，在一个受控的散热实验中，可以比较不同风速下的温度曲线，同时记录风扇耗电量。它要求的是关于改变条件后效果的证据，单靠原来的恢复曲线无法得到全部答案。

描述、解释、预测和干预可以围绕同一个对象展开，但各自需要的信息并不相同。系统科学让我们把组成和联系放在一起研究，也让我们把问题与证据对应起来。下一节将沿着这些问题，看看不同学术传统为什么发展出不同的方法。
