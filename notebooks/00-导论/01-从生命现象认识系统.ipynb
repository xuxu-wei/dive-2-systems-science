## 从一个熟悉的现象开始

快走或跑步结束后，我们可能仍然感觉热，心跳和呼吸也不会在停下的同一瞬间恢复。身体怎样完成这段恢复过程？

先聚焦其中一个可研究的问题：**运动结束后，热量怎样产生、在体内运输并向环境散失？** 肌肉活动改变产热，循环参与热量运输，皮肤与汗液蒸发参与散热；环境温度、湿度和空气流动会影响热量交换。调节过程又会随身体的情况改变。理解恢复，需要把这些过程放在一起。

这里涉及的生理调节关系可对照 [OpenStax 的稳态一节](https://openstax.org/books/anatomy-and-physiology-2e/pages/1-5-homeostasis)。下面的图是帮助提问的简化关系图，箭头表示考虑了哪类影响，不表示已经测得其大小。

## 什么是系统，什么是系统科学

**系统（system）**可以理解为相互关联、相互作用的组成部分所构成的整体。研究时还需要说明它与环境怎样联系，以及我们关心什么尺度上的行为。把器官名称列齐之后，仍需追问：谁影响谁，影响怎样传递，结果怎样随时间变化？

**系统科学（systems science）**研究系统的组成、相互作用、环境与整体行为之间的关系，并发展理解、预测和调节这些行为的方法。它把不同领域中的共同问题放在一起比较，也要求我们检查这种比较成立的条件。

一个恒温装置和人体都可以利用当前情况调整行为，这提供了共同研究调节的入口；人体的组成、约束与生理机制仍需另外研究。相似的关系帮助迁移方法，具体对象的知识帮助判断方法是否合适。

下图从左侧的活动，经中部的热量产生与运输，连接到右侧的散热过程。下方的环境和调节作用提示我们：同一项局部过程可能随外部条件和内部情况而改变。

| 图中英文 | 含义 |
|---|---|
| Activity / heat production | 活动与产热 |
| Internal heat / circulation | 内部热量与循环运输 |
| Skin / evaporation | 皮肤与蒸发散热 |
| Regulation | 身体调节 |
| Environment | 环境条件 |

```python
import matplotlib.pyplot as plt
from systems_science import plot_style

plot_style()
fig, ax = plt.subplots(figsize=(10, 4.5))
ax.set(xlim=(0, 10), ylim=(0, 5))
ax.axis('off')
positions = [(1.5, 3.5), (5, 3.5), (8.5, 3.5), (3.5, 1.1), (8.5, 1.1)]
labels = ['Activity /\nheat production', 'Internal heat /\ncirculation',
          'Skin /\nevaporation', 'Regulation', 'Environment']
boxes = []
for (x, y), label in zip(positions, labels):
    boxes.append(ax.text(x, y, label, ha='center', va='center', fontsize=12, zorder=3,
            bbox=dict(boxstyle='round,pad=0.6', facecolor='#ffffff', edgecolor='#6491b5')))
fig.canvas.draw()
# 箭头只表达本图纳入的联系；不赋予定量强度。
for source, target in [(0,1), (1,2), (1,3), (3,1), (3,2), (4,2)]:
    ax.annotate('', xy=positions[target], xytext=positions[source],
                arrowprops=dict(arrowstyle='->', color='#4078a0',
                                patchA=boxes[source].get_bbox_patch(),
                                patchB=boxes[target].get_bbox_patch(),
                                shrinkA=5, shrinkB=5, connectionstyle='arc3,rad=0.08'), zorder=2)
ax.set_title('A question about recovery: processes and their connections', pad=12)
plt.show()
```

## 把视角迁移到另外两个问题

物质进入身体之后，可能在不同组织间交换，再逐渐清除。某处读数下降，既可能涉及离开这一处，也可能涉及从整体清除。想研究局部与整体的变化，需要把交换关系和观察范围说清楚。这条线索将在第一至第三篇变成可以计算的模型（model）。

再看个体之间的消息传播。即使人数和联系数都相同，谁与谁联系仍然可能改变消息到达的顺序。六个人排成一条链，与一个人连接其余五个人，会发生同样的过程吗？第三节会从手工枚举开始回答它。

三个问题涉及的对象不同，却都需要研究组成、联系、环境与时间。这正是本书反复使用生命系统案例的原因：在具体问题中建立概念，再检查它能迁移到哪里。

| 研究目的 | 恢复问题中的例子 | 下一步需要什么 |
|---|---|---|
| 描述和解释 | 哪些过程共同影响恢复？ | 明确考虑的关系，收集可比较的观测（observation） |
| 预测 | 在给定活动和环境下，之后会怎样？ | 可以产生预测的模型，以及新的数据检验 |
| 干预 | 改变哪个条件有助于达到目标？ | 明确目标、可改变的条件、效果与代价 |

研究某个组成部分的机制，能够帮助我们写出整体模型；整体模型的偏差又能提示还缺少哪一项局部证据。例如，汗液分泌增加是否必然意味着散热同样增加，还要考察蒸发与环境的关系。两种研究角度因此形成来回推进的过程。

## 从现象走向下一步

系统科学首先改变的是提问方式：除了“有哪些部分”，也问“它们怎样联系、在什么条件下怎样变化、证据来自哪里”。之后，模型和计算让这些问题逐渐变得可以核对。
